```python
from graphviz import Digraph
import os

# Define node styles for reuse
node_styles = {
    'sensor': {'shape': 'box', 'style': 'filled', 'fillcolor': 'lightblue'},
    'interface': {'shape': 'box', 'style': 'filled', 'fillcolor': 'lightyellow'},
    'signal': {'shape': 'box', 'style': 'filled', 'fillcolor': 'lightgreen'},
    'protection': {'shape': 'box', 'style': 'filled', 'fillcolor': 'salmon'},
    'monitoring': {'shape': 'box', 'style': 'filled', 'fillcolor': 'plum'}
}

# Create a Digraph object with JPG format
dot = Digraph(format='jpg')
dot.attr(rankdir='LR', size='8')

# Function to add styled nodes
def add_node(name, label, style_key):
    style = node_styles.get(style_key, {})
    dot.node(name, label, **style)

# Define nodes
add_node('A1', 'Sensor Level\n(Prox, Vel, Accel,\nRTDs, Keyphasor)', 'sensor')
add_node('B1', 'Input Interface Module\n(Analog Preprocessing)', 'interface')
add_node('B2', 'A/D Conversion\n3500: in Monitor\nOrbit 60: Early Stage', 'interface')
add_node('C1', 'Signal Conditioning & Computation\n(RMS, Peak, 1X, FFT,\nGap, Smax, Order Tracking)', 'signal')
add_node('D1', 'Trip Logic\n(Alert/Danger,\nBypass, Voting,\nLatching, TMR)', 'protection')
add_node('E1', 'System 1 Export\nStatic (Gap, RMS...)\nDynamic (Waveform, FFT,\nOrbit, Waterfall)', 'monitoring')

# Define edges
edges = [('A1', 'B1'), ('B1', 'B2'), ('B2', 'C1'), ('C1', 'D1'), ('C1', 'E1')]
dot.edges(edges)

# Render the JPG file
output_filename = "turbomachinery_signal_path"
# output_dir = "/mnt/data"
# jpg_path = os.path.join(output_dir, f"{output_filename}.jpg")
dot.render(filename=output_filename, cleanup=True)

# jpg_path
```
